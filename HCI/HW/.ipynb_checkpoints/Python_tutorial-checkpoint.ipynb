```python
# import python libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import os

```

```python
print(os.getcwd())
```

```python
df = pd.read_csv("data.csv")
df.head(5)
```

```python
X = df['Size'].values.reshape(-1, 1)
y = df['Price']

reg = LinearRegression().fit(X, y)
print("slope :", reg.coef_[0], "intercept :", reg.intercept_)
```

```python
slope = reg.coef_[0]
intercept = reg.intercept_
y_pred = intercept + slope * df['Size']
```

```python
R2 = r2_score(y, y_pred)
#RMSE = mean_squared_error(y, y_pred, squared = False)
MSE = mean_squared_error(y, y_pred)
RMSE = np.sqrt(MSE)
```

```python
print("R\u00b2 value: ", R2)
```

```python
print("Root Mean Squared Error (RMSE): ", RMSE)
```

```python
plt.scatter(df['Size'], y, c = "tab:blue")
plt.plot(df['Size'], y_pred, c = "tab:red")
plt.title("Size vs. Price Regression", fontsize=20, pad=20)
plt.xlabel("Size", fontsize=16)
plt.ylabel("Price", fontsize=16)
plt.text(4000, 330000, r"$R^2$ : " + '{:.3f}'.format(R2))
plt.text(4000, 310000, f"y = {'{:.2f}'.format(intercept)} + {'{:.2f}'.format(slope)} $\cdot$ x")
plt.show()
```
